```python
import google.generativeai as genai
from crewai import Agent, Task, Crew
from litellm import completion  # Import LiteLLM for making requests to the model

# Configure the API key for Gemini
genai.configure(api_key="AIzaSyDqunZ1nqaLI53_0KKQsUERoZxNHyRLfeM")

# Define the function that uses LiteLLM completion with the correct model
def gemini_llm(prompt: str) -> str:
    try:
        # Direct call to LiteLLM's completion function
        response = completion(
            model="google/gemini-2.5-pro",  # Pass the correct model provider string
            messages=[{"role": "user", "content": prompt}]
        )
        return response['choices'][0]['message']['content'].strip()  # Return the model's response text
    except Exception as e:
        print(f"Error during Gemini API call: {str(e)}")
        return "Error during API call."

# Create CrewAI agents
researcher = Agent(
    role='Market Research Analyst',
    goal='Analyze competitors and summarize their marketing strategies',
    backstory='An expert in market intelligence and competitive analysis.',
    llm=gemini_llm,  # Pass the gemini_llm function as the LLM
    allow_delegation=False
)

writer = Agent(
    role='Content Strategist',
    goal='Use research to create a compelling marketing strategy document',
    backstory='A seasoned content strategist with a flair for storytelling.',
    llm=gemini_llm  # Pass the gemini_llm function as the LLM
)

# Define the tasks
task1 = Task(
    description="List top 3 competitors and their marketing strategies based on current trends.",
    agent=researcher,
    expected_output="A summary of 3 competitors with key marketing strategies."
)

task2 = Task(
    description="Create a content marketing strategy based on the competitor summary.",
    agent=writer,
    expected_output="A structured document with our content strategy inspired by competitors.",
    depends_on=[task1]
)

# Create Crew (the team)
crew = Crew(
    agents=[researcher, writer],
    tasks=[task1, task2],
    verbose=True  # Enable verbose output to debug agent behavior
)

# Run the Crew (Kickoff the task flow)
result = crew.kickoff()
print("Crew Result:", result)
```
