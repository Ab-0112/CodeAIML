#### finding missing values

```python
df.isnull().sum()
```

```python
df.rename(columns={
    "Total Revenue (B USD)": "Total_Revenue",
    "Net Income (B USD)": "Net_Income",
    "Total Assets (B USD)": "Total_Assets",
    "Total Liabilities (B USD)": "Total_Liabilities",
    "Cash Flow from Ops (B USD)": "Cash_Flow_Operations"
}, inplace=True)
```

```python
df.fillna(0)
```

#### Adding new columns

```python
df['Revenue Growth(%)'] = df.groupby(['Fiscal Year'])['Total_Revenue'].pct_change()*100
```

```python
df['Net Income Growth(%)'] = df.groupby(['Fiscal Year'])['Net_Income'].pct_change()*100
```
