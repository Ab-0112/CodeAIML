```python
import pandas as pd
flight = pd.read_excel('British Airways Summer Schedule Dataset - Forage Data Science Task 1.xlsx')
flight.head()
```

```python
flight.columns = flight.columns.str.lower()  # Set the column names to lowercase
```

```python
flight.columns
```

```python
sample_flights = flight[(flight['arrival_region'] == 'North America') & (flight['time_of_day'] == 'Morning')].copy()
```

```python
sample_flights.loc[:, 'category'] = sample_flights['time_of_day'] + ' - ' + sample_flights['arrival_region']
```

```python
tier1_pct = 0.8 
tier2_pct = 0.5 
tier3_pct = 0.2

sample_flights.loc[:, 'likely_lounge_users'] = (
    sample_flights['tier1_eligible_pax'] * tier1_pct +
    sample_flights['tier2_eligible_pax'] * tier2_pct +
    sample_flights['tier3_eligible_pax'] * tier3_pct
).astype(int)
```

```python
sample_flights1 = sample_flights[['category', 'arrival_station_cd','tier1_eligible_pax', 'tier2_eligible_pax', 'tier3_eligible_pax', 'likely_lounge_users']].head()
sample_flights1.head()
```

```python
sample_flights1.to_excel('sample_flights_export1.xlsx', index=True, sheet_name='Sample Flights', engine='openpyxl')
```
